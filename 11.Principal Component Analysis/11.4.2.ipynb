```python
import pandas as pd

data = {
    'Variable': [
        'Climate',
        'Housing',
        'Health',
        'Crime',
        'Transportation',
        'Education',
        'Arts',
        'Recreation',
        'Economy'
    ],
    
    'PC1': [
        0.190, 0.544, 0.782, 0.365, 0.585,
        0.394, 0.985, 0.520, 0.142
    ],
    
    'PC2': [
        0.017, 0.020, -0.605, 0.294, 0.085,
        -0.273, 0.126, 0.402, 0.150
    ],
    
    'PC3': [
        0.207, 0.204, 0.144, 0.585, 0.234,
        0.027, -0.111, 0.519, 0.239
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
for pc in ['PC1', 'PC2', 'PC3']:
    
    index = df[pc].abs().idxmax()
    
    variable = df.loc[index, 'Variable']
    loading = df.loc[index, pc]
    
    print(
        pc,
        "->",
        variable,
        "with loading",
        loading
    )
```
